# 25 Frozen daily versus 15:30 evaluation

Direct metric-matched intervals and a metadata-only audit of forward source coverage. This reader inspects saved results, does not fit models, and does not score a forward period. See evaluation_protocol.md for the frozen definitions and limitations.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Markdown
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'blackrock_data').exists())
out = root / 'outputs/experiments/25_frozen_intraday_evaluation'
display(Markdown((out / 'README.md').read_text()))
i = pd.read_csv(out / 'metric_intervals.csv')
display(i[i.dimension.eq('overall')])
display(pd.read_csv(out / 'forward_source_coverage.csv'))
display(i.groupby(['dimension', 'status']).size().rename('interval_records').reset_index())
checks = pd.read_csv(out / 'validation_checks.csv')
assert checks.passed.all()
assert json.loads((out / 'evaluation_summary.json').read_text())['forward_period_scored'] is False
assert json.loads((out / 'unused_period_audit.json').read_text())['forward_outcomes_scored'] is False
print(f'{len(checks)} checks passed; no forward period scored')
```
